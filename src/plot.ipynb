# RL-Mini-Project Results Visualization
This notebook visualizes the learning curves and final benchmark performance of the GNN-DQN and GNN-PPO models.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set seaborn theme for the entire notebook
sns.set_theme(style="darkgrid", context="paper", font_scale=1.2)
```

## 1. Learning Curve: Episode Return
Plotting the episode return over the training steps.

```python
df_learning = pd.read_csv("../results/learning_curve_data.csv")

plt.figure(figsize=(8, 5))

sns.lineplot(data=df_learning, x="step", y="episode_return", hue="method", 
             errorbar="sd", linewidth=2, palette=["blue", "orange"])

plt.title("Learning Curve: Episode Return")
plt.xlabel("Training Steps")
plt.ylabel("Mean Return (5 Eval Seeds)")

handles, labels = plt.gca().get_legend_handles_labels()
labels = [label.upper().replace("_", "-") for label in labels]
plt.legend(handles, labels)

plt.tight_layout()
plt.show()
```

## 2. Learning Constraints: Travel Cost & Late Deliveries
Visualizing how travel costs and late deliveries evolve during training.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Cost
sns.lineplot(data=df_learning, x="step", y="total_cost", hue="method", 
             errorbar="sd", ax=axes[0], palette=["blue", "orange"])
axes[0].set_title("Travel Cost")
axes[0].set_xlabel("Training Steps")
axes[0].set_ylabel("Mean Cost")
handles, labels = axes[0].get_legend_handles_labels()
labels = [label.upper().replace("_", "-") for label in labels]
axes[0].legend(handles, labels)

# Late
sns.lineplot(data=df_learning, x="step", y="n_late", hue="method", 
             errorbar="sd", ax=axes[1], palette=["blue", "orange"])
axes[1].set_title("Late Deliveries")
axes[1].set_xlabel("Training Steps")
axes[1].set_ylabel("Mean Late")
handles, labels = axes[1].get_legend_handles_labels()
labels = [label.upper().replace("_", "-") for label in labels]
axes[1].legend(handles, labels)

plt.tight_layout()
plt.show()
```

## 3. Benchmark Comparison
Comparing the final models against greedy and random baselines.

```python
df_bench = pd.read_csv("../results/eval_comparison.csv")
sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# 1. Return
sns.barplot(data=df_bench, x="method", y="episode_return", ax=axes[0], errorbar="sd", palette="viridis")
axes[0].set_title("Final Episode Return")
axes[0].set_ylabel("Return")

# 2. Cost
sns.barplot(data=df_bench, x="method", y="total_cost", ax=axes[1], errorbar="sd", palette="viridis")
axes[1].set_title("Total Travel Cost")
axes[1].set_ylabel("Cost")

# 3. On Time
sns.barplot(data=df_bench, x="method", y="n_on_time", ax=axes[2], errorbar="sd", palette="viridis")
axes[2].set_title("On-Time Deliveries")
axes[2].set_ylabel("Count")

for ax in axes:
    ax.set_xlabel("")
    labels = [item.get_text().upper().replace("_", "-") for item in ax.get_xticklabels()]
    ax.set_xticklabels(labels, rotation=30)

plt.tight_layout()
plt.show()
```
